# 🧬 PiFlow: Principle-aware Scientific Discovery Tutorial

Welcome to the PiFlow tutorial! This notebook is designed to guide scientists through using PiFlow to accelerate their research discovery process. 

**PiFlow** (Principle Flow) is an information-theoretical framework that treats automated scientific discovery as a structured uncertainty reduction problem, guided by foundational scientific principles. 

### 🔮 Why PiFlow?
- **Efficiency**: Up to 73% increase in discovery efficiency.
- **Rationality**: Hypotheses are guided by learned scientific principles, not just random search.
- **Plug-and-Play**: Easily adapt to Nanohelices, Bio-molecules, Superconductors, and more.
- **Flexibility**: Support all kinds of scenarios, such as materials discovery, drug design, and more.

## 🚀 Step 4: Running the Discovery Loop

Now we can run the discovery loop for the Nanohelix material discovery task.

The `inference.py` script starts the multi-agent system, which includes:
- **Planner**: Guides the exploration using scientific principles.
- **Hypothesis Agent**: Proposes new candidates (e.g., nanohelix structures).
- **Experiment Agent**: Validates candidates using tools (simulations).
- **User Proxy**: Represents the human user.

````python
# Start the simulation server in the background
import subprocess
import time

server_proc = subprocess.Popen(["python", "envs/start_server.py"])
time.sleep(5) # Wait for server to start

# Run the inference
!python inference.py --task_config configs/demo_config_for_task.yaml --model_config configs/demo_config_for_model.yaml --max_turn 5 --principled

# Terminate the server
server_proc.terminate()
````
